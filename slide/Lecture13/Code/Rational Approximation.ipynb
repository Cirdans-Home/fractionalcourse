```python
import numpy as np
import baryrat

alpha = 0.5
def f(x): return x**alpha
r = baryrat.brasil(f, [0,1], 5)
```

```python
Z = np.linspace(0.0, 1.0, 1000)
err = f(Z) - r(Z)
print(np.linalg.norm(err, np.inf))
```

```python
import matplotlib.pyplot as plt

alpha = 0.5
def f(x): return x**alpha
r = baryrat.brasil(f, [0,1], 5)
err = f(Z) - r(Z)
fig, ax = plt.subplots()
ax.plot(Z, err, linewidth=2.0)
ax.yscale = 'log'
plt.savefig('brasil5.pdf')
plt.show()
print(r.poles(0))
```
